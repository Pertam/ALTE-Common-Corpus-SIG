## Create project folder structure

Create the shared project folders and language-specific data folders.


```python
from pathlib import Path
import json

PROJECT_ROOT = Path('/content/drive/MyDrive/ALTE-Common-Corpus-SIG')

LANGUAGES = {
    'en': 'English',
    'fr': 'French',
    'es': 'Spanish',
    'de': 'German',
    'cs': 'Czech',
}

shared_folders = [
    'admin',
    'methodology',
    'data/raw',
    'data/interim',
    'data/outputs',
    'logs',
    'review',
    'exports',
    'secrets',
    'archive',
]

for folder in shared_folders:
    (PROJECT_ROOT / folder).mkdir(parents=True, exist_ok=True)

for lang_code, lang_name in LANGUAGES.items():
    folders = [
        f'data/raw/{lang_code}/corpus_original',
        f'data/raw/{lang_code}/metadata',
        f'data/raw/{lang_code}/licensing',
        f'data/interim/{lang_code}/tokens',
        f'data/interim/{lang_code}/lemma_sentence_index',
        f'data/interim/{lang_code}/lemma_stats',
        f'data/interim/{lang_code}/filtered_lemmas',
        f'data/interim/{lang_code}/samples',
        f'data/outputs/{lang_code}/pass1',
        f'data/outputs/{lang_code}/pass2',
        f'data/outputs/{lang_code}/pass3',
        f'data/outputs/{lang_code}/qa',
        f'data/outputs/{lang_code}/review_workbooks',
    ]

    for folder in folders:
        (PROJECT_ROOT / folder).mkdir(parents=True, exist_ok=True)

manifest = {
    'project_root': str(PROJECT_ROOT),
    'languages': LANGUAGES,
    'note': 'Created for ALTE Common Corpus SIG / European CEFR Vocabulary Atlas pilot.'
}

(PROJECT_ROOT / 'folder_manifest.json').write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding='utf-8'
)

print('Drive folder structure created at:', PROJECT_ROOT)
```
