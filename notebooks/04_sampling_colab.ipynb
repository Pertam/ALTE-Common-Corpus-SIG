# 04 Sampling

Use this notebook to sample lemmas and extract sentence examples for the LLM tagging stage.

## Methodological warning

This notebook is part of the ALTE Common Corpus SIG / European CEFR Vocabulary Atlas pilot workflow. Outputs produced by LLM-assisted steps are provisional Tier 4 candidate material only. They are not validated CEFR data and require expert review.


```python
from pathlib import Path
import os
import pandas as pd

REPO_ROOT = Path('/content/ALTE-Common-Corpus-SIG')
DRIVE_ROOT = Path('/content/drive/MyDrive/ALTE-Common-Corpus-SIG')

print('REPO_ROOT:', REPO_ROOT)
print('DRIVE_ROOT:', DRIVE_ROOT)
```

## Sampling settings

Current pilot setting: random 15 lemmas per language, all sentences for each selected lemma.


```python
N_LEMMAS_PER_LANGUAGE = 15
MIN_ARF_PER_MILLION = 50
LANGUAGES = ['en', 'fr', 'es', 'de', 'cs']

print('N_LEMMAS_PER_LANGUAGE:', N_LEMMAS_PER_LANGUAGE)
print('MIN_ARF_PER_MILLION:', MIN_ARF_PER_MILLION)
print('LANGUAGES:', LANGUAGES)
```

## Run sampling script

Call the sampling script once available.


```python
import subprocess
from pathlib import Path

# This cell runs Stage 04: random lemma selection plus sentence extraction.
# Current pilot setting:
#   - 15 lemmas per language
#   - ARF per million >= 50
#   - all available sentence examples for each selected lemma

INTERIM_DIR = DRIVE_ROOT / "data" / "interim"
STATS_DIR = DRIVE_ROOT / "data" / "stage3_lemma_stats"
SAMPLE_DIR = DRIVE_ROOT / "data" / "stage4_samples"

SAMPLE_DIR.mkdir(parents=True, exist_ok=True)

for lang in LANGUAGES:
    print("=" * 80)
    print("Sampling language:", lang)
    print("=" * 80)

    stats_path = STATS_DIR / f"{lang}_lemma_stats.csv"
    lemma_sentence_path = INTERIM_DIR / f"{lang}_lemma_sentence_index.parquet"
    sentences_path = INTERIM_DIR / f"{lang}_sentences.parquet"
    output_path = SAMPLE_DIR / f"stage4_{lang}_random_{N_LEMMAS_PER_LANGUAGE}_lemmas_all_sentences.csv"

    cmd = [
        "python",
        str(REPO_ROOT / "scripts" / "04_sample_lemmas_and_sentences.py"),
        "--lang", lang,
        "--stats", str(stats_path),
        "--lemma_sentence", str(lemma_sentence_path),
        "--sentences", str(sentences_path),
        "--output", str(output_path),
        "--min_arf", str(MIN_ARF_PER_MILLION),
        "--lemmas_n", str(N_LEMMAS_PER_LANGUAGE),
        "--all_sentences",
        "--seed", "20260603",
    ]

    print("Command:")
    print(" ".join(cmd))

    subprocess.run(cmd, check=True)

print("Stage 04 sampling complete.")
```
